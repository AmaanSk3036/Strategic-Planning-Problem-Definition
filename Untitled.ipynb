# EXPLORATORY DATA ANALYSIS (EDA)

# EXPLORE SALES PERFORMANCE 

```python

import matplotlib.pyplot as plt

# Group transactions by month and add their revenue
monthly_sales = (
    sales_df
    .set_index("InvoiceDate")["Revenue"]
    .resample("MS")
    .sum()
)

# Display the monthly revenue table
display(monthly_sales.to_frame(name="Revenue"))

# Draw a line chart of monthly revenue
monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(12, 5)
)

# Add chart labels and title
plt.title("Monthly Sales Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

# IDENTIFY BEST SELLING PRODUCT

```python

# Group transactions by product description
# and calculate total revenue for each product

product_sales = (
    sales_df.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

# Select the ten products with the highest revenue
top_10_products = product_sales.head(10)


# Display the results
display(top_10_products)


# Create a horizontal bar chart
top_10_products.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

# Add chart labels and title
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue (£)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

# COMPARE REVENUE ACROSS COUNTRIES 

```python

# Group sales by country and calculate total revenue
country_sales = (
    sales_df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

# Display the ten highest-revenue countries
display(country_sales.head(10))

# Plot the top ten countries
country_sales.head(10).sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Countries by Sales Revenue")
plt.xlabel("Revenue (£)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

# ANALYZE CUSTOMER PURCHASING BEHAVIOUR

```python

# Create a customer analysis dataset
# Exclude rows without a customer ID for this analysis only
customer_df = sales_df.dropna(
    subset=["CustomerID"]
).copy()

# Calculate total revenue for each customer
customer_revenue = (
    customer_df.groupby("CustomerID")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

# Count the number of distinct invoices per customer
customer_orders = (
    customer_df.groupby("CustomerID")["InvoiceNo"]
    .nunique()
)

# Combine revenue and order counts into one table
customer_summary = pd.DataFrame({
    "TotalRevenue": customer_revenue,
    "NumberOfOrders": customer_orders
})

# Display the ten customers with the highest revenue
display(customer_summary.head(10))

# Measure the correlation between order count and revenue
correlation = customer_summary[
    ["NumberOfOrders", "TotalRevenue"]
].corr().iloc[0, 1]

print("Correlation:", correlation)
```

# DATA VISUALIZATION 

# PREPARE DATA FOR VISUALIZATION 

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a fresh copy so the original data is not modified
eda_df = df.copy()

# Convert invoice date into a proper datetime format
eda_df["InvoiceDate"] = pd.to_datetime(
    eda_df["InvoiceDate"], errors="coerce"
)

# Convert quantity and price into numeric values
eda_df["Quantity"] = pd.to_numeric(
    eda_df["Quantity"], errors="coerce"
)
eda_df["UnitPrice"] = pd.to_numeric(
    eda_df["UnitPrice"], errors="coerce"
)

# Calculate the revenue of each transaction line
eda_df["Revenue"] = eda_df["Quantity"] * eda_df["UnitPrice"]

# Create a column identifying cancellation-marked invoices
eda_df["IsCancellation"] = (
    eda_df["InvoiceNo"].astype("string").str.startswith(
        "C", na=False
    )
)

# Create a dataset for positive-quantity, positive-price sales
# Keep the original eda_df for investigating returns and cancellations
sales_df = eda_df.loc[
    eda_df["InvoiceDate"].notna()
    & eda_df["Quantity"].gt(0)
    & eda_df["UnitPrice"].gt(0)
].copy()

# Extract year, month, and weekday for time-based analysis
sales_df["YearMonth"] = (
    sales_df["InvoiceDate"].dt.to_period("M").astype(str)
)
sales_df["Weekday"] = sales_df["InvoiceDate"].dt.day_name()
sales_df["Hour"] = sales_df["InvoiceDate"].dt.hour

# Create a unique invoice-level dataset for order analysis
orders_df = (
    sales_df.groupby("InvoiceNo")
    .agg(
        OrderValue=("Revenue", "sum"),
        TotalItems=("Quantity", "sum"),
        OrderDate=("InvoiceDate", "min"),
        Country=("Country", "first")
    )
    .reset_index()
)

print("Rows available for sales EDA:", len(sales_df))
print("Distinct invoices:", sales_df["InvoiceNo"].nunique())
print("Distinct identified customers:",
      sales_df["CustomerID"].nunique())
```

# SALES TREND ANALYSIS 

```python

# Calculate total revenue for each month
monthly_revenue = (
    sales_df.groupby("YearMonth")["Revenue"].sum()
)

# Create the line chart
plt.figure(figsize=(13, 5))
plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Sales Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=60)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Display the highest-revenue month in the available data
print("Highest revenue month:",
      monthly_revenue.idxmax())
print("Revenue in that month:",
      round(monthly_revenue.max(), 2))
```

# COMPARE SALES BY DAY OF WEEK 

```python

# Define the correct weekday order
weekday_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

# Calculate revenue for each weekday
weekday_revenue = (
    sales_df.groupby("Weekday")["Revenue"]
    .sum()
    .reindex(weekday_order, fill_value=0)
)

# Plot weekday revenue
plt.figure(figsize=(10, 5))
weekday_revenue.plot(kind="bar")

plt.title("Total Revenue by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()

display(weekday_revenue.to_frame("Revenue"))
```

# PRODUCT PERFORMANCE - REVENUE VS QUANTITY SOLD 

```python

# Group products by description
product_summary = (
    sales_df.groupby("Description")
    .agg(
        QuantitySold=("Quantity", "sum"),
        Revenue=("Revenue", "sum"),
        NumberOfOrders=("InvoiceNo", "nunique")
    )
    .dropna()
)

# Keep products with positive total quantity and revenue
product_summary = product_summary.loc[
    (product_summary["QuantitySold"] > 0)
    & (product_summary["Revenue"] > 0)
]

# Draw the scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(
    product_summary["QuantitySold"],
    product_summary["Revenue"],
    alpha=0.5
)

plt.title("Product Quantity Sold vs Revenue")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Revenue (£)")
plt.tight_layout()
plt.show()

# Display products with the highest revenue
display(
    product_summary.sort_values(
        "Revenue", ascending=False
    ).head(10)
)
```

# ORDER VALUE ANALYSIS - HOW MUCH DO CUSTOMERS SPEND PER ORDER ?

```python

# Each row of orders_df represents one invoice
order_values = orders_df["OrderValue"]

# Calculate descriptive statistics for order values
print("Average order value:",
      round(order_values.mean(), 2))
print("Median order value:",
      round(order_values.median(), 2))
print("Maximum order value:",
      round(order_values.max(), 2))

# Plot the distribution, excluding non-positive order values
# from this particular chart
positive_order_values = order_values[order_values > 0]

plt.figure(figsize=(10, 5))
plt.hist(positive_order_values, bins=50)

plt.title("Distribution of Order Values")
plt.xlabel("Order Value (£)")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

# CUSTOMER ANALYSIS - WHO CONTRIBUTES THE MOST REVENUE ?

```python

# CustomerID is required for customer-level analysis
identified_sales = sales_df.dropna(
    subset=["CustomerID"]
).copy()

# Calculate revenue and distinct invoice count for each customer
customer_summary = (
    identified_sales.groupby("CustomerID")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        NumberOfOrders=("InvoiceNo", "nunique"),
        TotalQuantity=("Quantity", "sum")
    )
    .sort_values("TotalRevenue", ascending=False)
)

# Select the top 10 customers
top_customers = customer_summary.head(10)

# Draw the horizontal bar chart
plt.figure(figsize=(10, 6))
plt.barh(
    top_customers.index.astype(str)[::-1],
    top_customers["TotalRevenue"].values[::-1]
)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Total Revenue (£)")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()

display(top_customers.round(2))
```

# COMPARE ORDER FREQUENCY AND SPENDING 

```python

# Use the customer summary calculated in the previous section
plt.figure(figsize=(10, 6))

plt.scatter(
    customer_summary["NumberOfOrders"],
    customer_summary["TotalRevenue"],
    alpha=0.5
)

plt.title("Customer Order Frequency vs Total Revenue")
plt.xlabel("Number of Distinct Orders")
plt.ylabel("Total Customer Revenue (£)")
plt.tight_layout()
plt.show()

# Calculate Pearson correlation between the two variables
customer_correlation = customer_summary[
    ["NumberOfOrders", "TotalRevenue"]
].corr().iloc[0, 1]

print("Correlation:", round(customer_correlation, 3))
```

# COMPARE REVENUE & ORDER VALUE BY MONTH 

```python

# Group transaction lines by month to calculate revenue
monthly_revenue = (
    sales_df.groupby("YearMonth")["Revenue"].sum()
)

# Group invoice-level records by month to count orders
orders_df["YearMonth"] = (
    orders_df["OrderDate"].dt.to_period("M").astype(str)
)

monthly_orders = (
    orders_df.groupby("YearMonth")["InvoiceNo"].nunique()
)

# Combine monthly metrics into a single table
monthly_summary = pd.DataFrame({
    "Revenue": monthly_revenue,
    "Orders": monthly_orders
}).fillna(0)

# Plot the two metrics separately because they use different units
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].plot(
    monthly_summary.index,
    monthly_summary["Revenue"],
    marker="o"
)
axes[0].set_title("Monthly Revenue")
axes[0].set_ylabel("Revenue (£)")
axes[0].tick_params(axis="x", rotation=60)
axes[0].grid(True, alpha=0.3)

axes[1].plot(
    monthly_summary.index,
    monthly_summary["Orders"],
    marker="o"
)
axes[1].set_title("Monthly Number of Orders")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Number of Orders")
axes[1].tick_params(axis="x", rotation=60)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# RETURN & CANCELLATION ANALYSIS 

```python

# Work from the full dataset, not just positive sales
cancelled_lines = eda_df.loc[
    eda_df["IsCancellation"]
    & eda_df["InvoiceDate"].notna()
].copy()

# Count cancellation-marked transaction lines by month
cancelled_lines["YearMonth"] = (
    cancelled_lines["InvoiceDate"]
    .dt.to_period("M").astype(str)
)

monthly_cancellations = (
    cancelled_lines.groupby("YearMonth").size()
)

# Plot the monthly count
plt.figure(figsize=(12, 5))
monthly_cancellations.plot(
    kind="line", marker="o"
)

plt.title("Cancellation-Marked Transaction Lines by Month")
plt.xlabel("Month")
plt.ylabel("Number of Transaction Lines")
plt.xticks(rotation=60)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Cancellation-marked transaction lines:",
      len(cancelled_lines))
print("Distinct cancellation-marked invoices:",
      cancelled_lines["InvoiceNo"].nunique())
```
